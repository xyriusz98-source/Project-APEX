## ANALYSIS

Liczymy cenę za m² raz dla każdego mieszkania (żeby nie przeliczać jej
wielokrotnie w kolejnych funkcjach) oraz przygotowujemy średnie potrzebne
do reguł biznesowych.

````python
def calculate_price_per_m2(price: int, area: int) -> float:
    """Zwraca cenę za m2. Rzuca ValueError, jeśli area <= 0."""
    if area <= 0:
        raise ValueError(f"Nieprawidłowa powierzchnia: {area}")
    return price / area


def enrich_with_price_per_m2(apartments: list[Apartment]) -> list[dict]:
    """Liczy price_per_m2 RAZ dla każdego mieszkania i dokleja jako pole,
    żeby nie przeliczać tej samej wartości w wielu miejscach."""
    enriched = []
    for apartment in apartments:
        record = dict(apartment)
        record["price_per_m2"] = calculate_price_per_m2(
            apartment["price"], apartment["area"]
        )
        enriched.append(record)
    return enriched


def calculate_average_price_per_m2(apartments: list[dict]) -> float:
    """Średnia cena/m2 po wszystkich mieszkaniach."""
    return mean(a["price_per_m2"] for a in apartments)


def average_price_per_m2_by_location(apartments: list[dict]) -> dict[str, float]:
    """Średnia cena/m2 w rozbiciu na lokalizacje."""
    by_location: dict[str, list[float]] = {}
    for a in apartments:
        by_location.setdefault(a["location"], []).append(a["price_per_m2"])
    return {loc: mean(vals) for loc, vals in by_location.items()}


enriched = enrich_with_price_per_m2(apartments)
avg_ppm2 = calculate_average_price_per_m2(enriched)

print(f"Średnia cena/m2 (cały rynek): {avg_ppm2:.2f} PLN/m2")
print("Średnia cena/m2 wg lokalizacji:")
for location, avg in average_price_per_m2_by_location(enriched).items():
    print(f"  {location}: {avg:.2f} PLN/m2")
````

## BUSINESS RULES

**Attractiveness Score** — pomocniczy, ciągły wynik do sortowania (Zadanie 4).

**APEX Score** — regułowy system ekspercki (Zadanie 5), 0-4 punkty za:
* cenę/m² poniżej średniej,
* powierzchnię > 50 m²,
* posiadanie balkonu,
* lokalizację uznaną za "dobrą".

Założenie biznesowe: lokalizacja jest **"dobra"**, jeśli jej średnia
cena/m² przewyższa średnią rynkową — czyli rynek już wycenia ją jako
bardziej pożądaną. To jedna z możliwych definicji — jeśli chcesz inną
(np. bliskość centrum, dostęp do komunikacji), to jest miejsce do zmiany.

````python
# Wagi scoringu atrakcyjności — jawnie nazwane stałe zamiast magicznych liczb
ATTRACTIVENESS_PRICE_WEIGHT = 100_000  # skala 1/price_per_m2
ATTRACTIVENESS_BALCONY_BONUS = 100
ATTRACTIVENESS_FLOOR_WEIGHT = 20
ATTRACTIVENESS_ROOM_WEIGHT = 30


def calculate_attractiveness_score(apartment: dict) -> float:
    """Ciągły wynik atrakcyjności (do sortowania/rankingu)."""
    score = ATTRACTIVENESS_PRICE_WEIGHT / apartment["price_per_m2"]
    if apartment["balcony"]:
        score += ATTRACTIVENESS_BALCONY_BONUS
    score += apartment["floor"] * ATTRACTIVENESS_FLOOR_WEIGHT
    score += apartment["rooms"] * ATTRACTIVENESS_ROOM_WEIGHT
    return score


def determine_good_locations(apartments: list[dict]) -> set[str]:
    """Lokalizacja jest 'dobra', jeśli jej średnia cena/m2 przewyższa
    średnią rynkową."""
    overall_avg = calculate_average_price_per_m2(apartments)
    location_avgs = average_price_per_m2_by_location(apartments)
    return {loc for loc, avg in location_avgs.items() if avg > overall_avg}


def calculate_apex_score(apartment: dict, avg_price_per_m2: float, good_locations: set[str]) -> int:
    """Regułowy scoring ekspercki (Zadanie 5). 0-4 punkty."""
    score = 0
    if apartment["price_per_m2"] < avg_price_per_m2:
        score += 1
    if apartment["area"] > 50:
        score += 1
    if apartment["balcony"]:
        score += 1
    if apartment["location"] in good_locations:
        score += 1
    return score


good_locations = determine_good_locations(enriched)
print(f"Lokalizacje uznane za 'dobre': {good_locations}")
````

## REPORT

Trzy osobne rankingi TOP 5 (cena/m², powierzchnia, atrakcyjność) oraz
pełna lista wyników APEX Score dla wszystkich 30 mieszkań.

````python
def rank_by_price_per_m2(apartments: list[dict], top_n: int = 5) -> list[dict]:
    """Ranking wg ceny/m2 — im niżej, tym lepiej (okazja cenowa)."""
    return sorted(apartments, key=lambda a: a["price_per_m2"])[:top_n]


def rank_by_area(apartments: list[dict], top_n: int = 5) -> list[dict]:
    """Ranking wg powierzchni — im więcej, tym lepiej."""
    return sorted(apartments, key=lambda a: a["area"], reverse=True)[:top_n]


def rank_by_attractiveness(apartments: list[dict], top_n: int = 5) -> list[dict]:
    """Ranking wg złożonego wyniku atrakcyjności."""
    return sorted(apartments, key=calculate_attractiveness_score, reverse=True)[:top_n]


def print_ranking(title: str, ranked: list[dict]) -> None:
    print()
    print(title)
    for position, apartment in enumerate(ranked, start=1):
        name = apartment["name"]
        location = apartment["location"]
        area = apartment["area"]
        price = apartment["price"]
        price_per_m2 = apartment["price_per_m2"]
        print(f"  {position}. {name} ({location}, {area} m2, {price} PLN, {price_per_m2:.0f} PLN/m2)")


print_ranking("TOP 5 - najniższa cena/m2:", rank_by_price_per_m2(enriched))
print_ranking("TOP 5 - najwieksza powierzchnia:", rank_by_area(enriched))
print_ranking("TOP 5 - najwyższa atrakcyjnosc:", rank_by_attractiveness(enriched))
````

````python
def print_apex_scores(apartments: list[dict], avg_price_per_m2: float, good_locations: set[str]) -> None:
    print("APEX Score dla kazdego mieszkania:")
    for apartment in apartments:
        score = calculate_apex_score(apartment, avg_price_per_m2, good_locations)
        name = apartment["name"]
        building = apartment["building"]
        print(f"  {name} (bud. {building}): APEX Score = {score}")


print_apex_scores(enriched, avg_ppm2, good_locations)
````
